```python
# Task 3: Handling Null Values
# Create a new DataFrame with some null values and fill them

# Your code here:


try:
    # Create a DataFrame containing null values
    df_nulls = pd.DataFrame({
        'Score': [88, np.nan, 92.0, 79.0],
        'Bonus': [500.0, 300.0, np.nan, 400.0],
        'Grade': ['B', 'A', 'C', None]
    })

    # Display the original DataFrame
    print("Original DataFrame:")
    print(df_nulls)

    # Count missing values in each column
    print("\nMissing Values:")
    print(df_nulls.isnull().sum())

    # Method 1: Fill numeric nulls with 0
    df_filled_zero = df_nulls.copy()

    numeric_columns = df_filled_zero.select_dtypes(include=[np.number]).columns

    df_filled_zero[numeric_columns] = (df_filled_zero[numeric_columns].fillna(0))

    # Fill text nulls with 'Missing'
    text_columns = df_filled_zero.select_dtypes(include=['object', 'string']).columns

    df_filled_zero[text_columns] = (df_filled_zero[text_columns].fillna('Missing'))

    print("\nMethod 1: Fill with Zero:")
    print(df_filled_zero)

    # Method 2: Fill numeric nulls with column mean
    df_filled_mean = df_nulls.copy()

    for col in numeric_columns:
        mean_value = df_filled_mean[col].mean()
        df_filled_mean[col] = (df_filled_mean[col].fillna(mean_value))

    # Fill text nulls with 'Unknown'
    df_filled_mean[text_columns] = (df_filled_mean[text_columns].fillna('Unknown'))

    print("\nMethod 2: Fill with Mean:")
    print(df_filled_mean)

    # Verify that no null values remain
    print("\nRemaining Null Values:")
    print(df_filled_mean.isnull().sum())

except Exception as e:
    print("Error:", e)
```

```python
from IPython.terminal.embed import line_magic
# Task 9: Iterators and Generators
# Create an iterator for the first 5 even numbers.
# Then, create a generator for the same.

# Your code here:

class FirstFiveEvenNumbers:
    def __init__(self, limit = 5):
      if limit < 0:
        raise ValueError("Limit must be a positive integer.")
      self.limit = limit
      self.current_number = 0
      self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.limit:
            raise StopIteration
        value = self.current_number
        self.current_number += 2
        self.count += 1
        return value

def even_numbers_generator(limit = 5):
    if limit < 0:
        raise ValueError("Limit must be a positive integer.")
    current_number = 0
    for _  in range(limit):
        yield current_number
        current_number += 2

even_iterator = FirstFiveEvenNumbers(5)
print(list(even_iterator))

even_generator = even_numbers_generator(5)
print(list(even_generator))
```

```python
# Task 10: Map, Reduce, and Filter
# Use map to square all numbers in a list.
# Use reduce to find the product of all numbers in a list.
# Use filter to get only even numbers from a list.

numbers = [1, 2, 3, 4, 5]

# Your code here:


# Task 10: Map, Reduce, and Filter

from functools import reduce

numbers = [1, 2, 3, 4, 5]

try:
    # Use map() to square all numbers
    squared_numbers = list(map(lambda x: x ** 2, numbers))
    print("Squared Numbers:", squared_numbers)

    # Use reduce() to calculate the product of all numbers
    product = reduce(lambda x, y: x * y, numbers, 1)
    print("Product of Numbers:", product)

    # Use filter() to extract even numbers
    even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
    print("Even Numbers:", even_numbers)

except Exception as e:
    print("Error:", e)
```

```python
# Task 11: Object-Oriented Programming - Creating a Class
# Create a class called 'Rectangle' with attributes 'length' and 'width'.
# Include methods to calculate area and perimeter.

# Your code here:


# Task 11: Object-Oriented Programming

try:
    # Create a Rectangle class
    class Rectangle:

        # Initialize length and width
        def __init__(self, length, width):
            if length <= 0 or width <= 0:
                raise ValueError("Length and width must be positive.")

            self.length = length
            self.width = width

        # Calculate the area
        def area(self):
            return self.length * self.width

        # Calculate the perimeter
        def perimeter(self):
            return 2 * (self.length + self.width)

    # Create two Rectangle objects
    rectangle1 = Rectangle(10, 5)
    rectangle2 = Rectangle(8, 4)

    # Display results for Rectangle 1
    print("Rectangle 1:")
    print("Area:", rectangle1.area())
    print("Perimeter:", rectangle1.perimeter())

    # Display results for Rectangle 2
    print("\nRectangle 2:")
    print("Area:", rectangle2.area())
    print("Perimeter:", rectangle2.perimeter())

except Exception as e:
    print("Error:", e)
```
