```python
# Python and Pandas Assessment Notebook

import pandas as pd
import numpy as np
from functools import reduce

# Task 1: Create a DataFrame using Pandas
# Create a DataFrame with the following data:
# Name: Alice, Bob, Charlie, David
# Age: 25, 30, 35, 28
# City: New York, San Francisco, Los Angeles, Chicago

# Your code here:

try:
    # Create a DataFrame with Name, Age, and City
    df = pd.DataFrame({
        'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Age': [25, 30, 35, 28],
        'City': ['New York', 'San Francisco', 'Los Angeles', 'Chicago']
    })

    # Convert columns to appropriate data types
    df['Name'] = df['Name'].astype(str)
    df['Age'] = df['Age'].astype(int)
    df['City'] = df['City'].astype(str)

    # Display the DataFrame
    print(df)

    # Verify the data types
    print("\nData Types:")
    print(df.dtypes)

except Exception as e:
    print("Error:", e)
```

```python
# Task 3: Handling Null Values
# Create a new DataFrame with some null values and fill them

# Your code here:


try:
    # Create a DataFrame containing null values
    df_nulls = pd.DataFrame({
        'Score': [88, np.nan, 92.0, 79.0],
        'Bonus': [500.0, 300.0, np.nan, 400.0],
        'Grade': ['B', 'A', 'C', None]
    })

    # Display the original DataFrame
    print("Original DataFrame:")
    print(df_nulls)

    # Count missing values in each column
    print("\nMissing Values:")
    print(df_nulls.isnull().sum())

    # Method 1: Fill numeric nulls with 0
    df_filled_zero = df_nulls.copy()

    numeric_columns = df_filled_zero.select_dtypes(include=[np.number]).columns

    df_filled_zero[numeric_columns] = (df_filled_zero[numeric_columns].fillna(0))

    # Fill text nulls with 'Missing'
    text_columns = df_filled_zero.select_dtypes(include=['object', 'string']).columns

    df_filled_zero[text_columns] = (df_filled_zero[text_columns].fillna('Missing'))

    print("\nMethod 1: Fill with Zero:")
    print(df_filled_zero)

    # Method 2: Fill numeric nulls with column mean
    df_filled_mean = df_nulls.copy()

    for col in numeric_columns:
        mean_value = df_filled_mean[col].mean()
        df_filled_mean[col] = (df_filled_mean[col].fillna(mean_value))

    # Fill text nulls with 'Unknown'
    df_filled_mean[text_columns] = (df_filled_mean[text_columns].fillna('Unknown'))

    print("\nMethod 2: Fill with Mean:")
    print(df_filled_mean)

    # Verify that no null values remain
    print("\nRemaining Null Values:")
    print(df_filled_mean.isnull().sum())

except Exception as e:
    print("Error:", e)
```

```python
# Task 4: GroupBy and Describe
# Using the following DataFrame, group by 'Category' and describe the 'Value' column

df_group = pd.DataFrame({
    'Category': ['A', 'B', 'A', 'B', 'A', 'C'],
    'Value': [10, 20, 15, 25, 30, 35]
})

# Your code here:

try:
    # Create the DataFrame
    grouped_stats = df_group.groupby('Category')['Value'].describe()

    # Display the results
    print(grouped_stats)

    for cat, stats in grouped_stats.iterrows():
        print(f"Category: {cat}")
        print(f"Count: {stats['count']}")
        print(f"Mean: {stats['mean']}")
        print(f"Std: {stats['std']}")

except Exception as e:
    print("Error:", e)
```
